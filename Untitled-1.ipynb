```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn

data = pd.read_csv(r'')
data.head(10)

data.drop_duplicates()


year = np.array([2020, 2021, 2022, 2023, 2024, 2025, 2026])
salesman = np.array([100, 300, 600, 200, 890, 700, 1000])
salesman2 = np.array([50, 100, 500, 800, 990, 451, 2000])
salesman3 = np.array([20, 850, 450, 900, 1000, 320, 800])

plt.tick_params(colors='black', axis='both')
plt.title("Year", color='red', fontweight='30')
plt.plot(year, salesman, label='Marcos')
plt.plot(year, salesman2, label='Gustavo')
plt.plot(year, salesman3, label='Larissa')
plt.xlabel(color='red', fontweight='20', family='Arial')
plt.ylabel(color='green', fontweight='20', family='Arial')
plt.xticks(year)
plt.legend()
plt.show()


```
